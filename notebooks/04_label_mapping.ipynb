```python
import re
import wfdb

def extract_patient_metadata(record_path):
    
    record = wfdb.rdheader(record_path)
    
    comments = " ".join(record.comments)
    
    age_match = re.search(r"Age:\s*(\d+)", comments)
    sex_match = re.search(r"Sex:\s*([A-Za-z]+)", comments)
    nyha_match = re.search(
        r"NYHA class:\s*([IVX]+(?:-[IVX]+)?)",
        comments,
        re.IGNORECASE
    )
    
    age = int(age_match.group(1)) if age_match else np.nan
    sex = sex_match.group(1) if sex_match else None
    nyha = nyha_match.group(1) if nyha_match else None
    
    return {
        "age": age,
        "sex": sex,
        "nyha_class": nyha
    }
```

```python
print(
    extract_patient_metadata(
        "../data/raw/chfdb/chf01"
    )
)

print(
    extract_patient_metadata(
        "../data/raw/chf2db/chf201"
    )
)
```

```python
metadata_records = []

# CHF2DB
for i in range(201, 230):
    
    patient_id = f"chf{i}"
    
    record_path = f"../data/raw/chf2db/{patient_id}"
    
    metadata = extract_patient_metadata(record_path)
    
    metadata["patient_id"] = patient_id
    metadata["database"] = "CHF2DB"
    
    metadata_records.append(metadata)


# CHFDB
for i in range(1, 16):
    
    patient_id = f"chf{i:02d}"
    
    record_path = f"../data/raw/chfdb/{patient_id}"
    
    metadata = extract_patient_metadata(record_path)
    
    metadata["patient_id"] = patient_id
    metadata["database"] = "CHFDB"
    
    metadata_records.append(metadata)


metadata_df = pd.DataFrame(metadata_records)

print(metadata_df)
```

```python
def assign_risk_label(nyha_class):
    
    if nyha_class in ["I", "II"]:
        return 0
    
    elif nyha_class in ["III", "III-IV"]:
        return 1
    
    else:
        return np.nan


metadata_df["risk_label"] = (
    metadata_df["nyha_class"]
    .apply(assign_risk_label)
)

print(
    metadata_df[
        ["patient_id", "nyha_class", "risk_label"]
    ]
)
```
